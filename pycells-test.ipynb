```python
import math, sys

print(f"python {sys.version_info.major}.{sys.version_info.minor}")

def area(r: float) -> float:
    """Circle area — a tiny function to show definitions persist between cells."""
    return math.pi * r**2

f"{area(2):.4f}"
```

```python
# uses area() from the previous cell — state persists across cells
[f"{area(r):.2f}" for r in (1, 2, 3)]
```
